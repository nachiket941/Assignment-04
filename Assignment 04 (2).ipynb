```python
# Write a Python program to check whether a number is even or odd.

num = int(input("Enter a number: "))

if num % 2 == 0:
    print("Even number")
else:
    print("Odd number")
```

```python
# Write a program to find the largest of three numbers.

a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
c = int(input("Enter third number: "))

if a >= b and a >= c:
    print("Largest number is:", a)
elif b >= a and b >= c:
    print("Largest number is:", b)
else:
    print("Largest number is:", c)
```

```python
# Write a program to check whether a given number is prime.

num = int(input("Enter a number: "))

if num <= 1:
    print("Not a prime number")
else:
    for i in range(2, num):
        if num % i == 0:
            print("Not a prime number")
            break
    else:
        print("Prime number")
```

```python
# Write a program to print the Fibonacci series up to n terms.

n = int(input("Enter number of terms: "))

a = 0
b = 1

for i in range(n):
    print(a, end=" ")
    a, b = b, a + b
```

```python
# Write a program to find the factorial of a number using a loop.

num = int(input("Enter a number: "))

fact = 1

for i in range(1, num + 1):
    fact = fact * i

print("Factorial is:", fact)
```

```python
# Write a program to reverse a string without using [::-1].

text = input("Enter a string: ")

reverse = ""

for i in text:
    reverse = i + reverse

print("Reversed string:", reverse)
```

```python
# Write a program to check whether a string is a palindrome.

text = input("Enter a string: ")

reverse = ""

for i in text:
    reverse = i + reverse

if text == reverse:
    print("Palindrome")
else:
    print("Not a palindrome")
```

```python
# Write a program to count the number of vowels and consonants in a string.

text = input("Enter a string: ")

vowels = 0
consonants = 0

for i in text:
    if i.isalpha():
        if i.lower() in "aeiou":
            vowels += 1
        else:
            consonants += 1

print("Vowels:", vowels)
print("Consonants:", consonants)
```

```python
# Write a program to find the sum of all elements in a list.

numbers = [10, 20, 30, 40, 50]

total = 0

for i in numbers:
    total = total + i

print("Sum of list:", total)
```

```python
# Write a program to find the largest and smallest element in a list without using max() or min().

numbers = [10, 25, 5, 40, 15]

largest = numbers[0]
smallest = numbers[0]

for i in numbers:
    if i > largest:
        largest = i

    if i < smallest:
        smallest = i

print("Largest:", largest)
print("Smallest:", smallest)
```

```python
# Write a program to remove duplicate elements from a list.

numbers = [10, 20, 10, 30, 20, 40]

new_list = []

for i in numbers:
    if i not in new_list:
        new_list.append(i)

print("List after removing duplicates:", new_list)
```

```python
# Write a program to count how many times each element appears in a list using a dictionary.

numbers = [10, 20, 10, 30, 20, 10]

count = {}

for i in numbers:
    if i in count:
        count[i] += 1
    else:
        count[i] = 1

print("Element frequency:", count)
```

```python
# Write a program to find the second-largest number in a list.

numbers = [10, 50, 30, 40, 20]

largest = second = float('-inf')

for i in numbers:
    if i > largest:
        second = largest
        largest = i
    elif largest > i > second:
        second = i

if second == float('-inf'):
    print("Second largest number does not exist")
else:
    print("Second largest number:", second)
```

```python
# Write a program to sort a list without using sort() or sorted().

numbers = [40, 10, 30, 20, 50]

for i in range(len(numbers)):
    for j in range(i + 1, len(numbers)):
        if numbers[i] > numbers[j]:
            numbers[i], numbers[j] = numbers[j], numbers[i]

print("Sorted list:", numbers)
```

```python
# Write a function that accepts a list of numbers and returns a list containing only the even numbers.

def even_numbers(numbers):
    result = []

    for i in numbers:
        if i % 2 == 0:
            result.append(i)

    return result

numbers = [10, 15, 20, 25, 30]

print("Even numbers:", even_numbers(numbers))
```

```python
# Write a program to find the frequency of each character in a string.

text = input("Enter a string: ")

count = {}

for i in text:
    if i in count:
        count[i] += 1
    else:
        count[i] = 1

print("Character frequency:", count)
```

```python
# Write a program to check whether two strings are anagrams of each other.

str1 = input("Enter first string: ")
str2 = input("Enter second string: ")

if sorted(str1.lower()) == sorted(str2.lower()):
    print("Strings are anagrams")
else:
    print("Strings are not anagrams")
```

```python
# Write a program to find all duplicate values in a list.

numbers = [10, 20, 30, 10, 40, 20, 50]

duplicates = []

for i in numbers:
    if numbers.count(i) > 1 and i not in duplicates:
        duplicates.append(i)

print("Duplicate values:", duplicates)
```

```python
# Write a program that takes a sentence and finds the longest word.

sentence = input("Enter a sentence: ")

words = sentence.split()

longest = ""

for i in words:
    if len(i) > len(longest):
        longest = i

print("Longest word:", longest)
```
